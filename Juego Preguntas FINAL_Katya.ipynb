```python
def jugar_trivial():
    print("🎲 Bienvenido al juego Trivial 🎲\n")

    # Inicialización de variables
    acertadas = 0
    fallos = 0
    contestadas = 0

    random.shuffle(preguntas)  # escoge una pregunta aleatoriamente y no repite.

    i = 0  # índice o posicion de la lista de preguntas.

    # Mientras no gane ni pierda y haya preguntas sigue el bucle.
    # para interrumpir el juego escribe STOP.
    while acertadas < 5 and fallos < 3 and i < len(preguntas):
        p = preguntas[i]

        print(f"Pregunta {i + 1}: {p['pregunta']}")
        for opcion in p["opciones"]:
            print(opcion)

        respuesta_usuario = input("Tu respuesta (A, B, C o D): ").upper()
        contestadas += 1

        if respuesta_usuario == "STOP":
            print("⏹ Juego detenido por el jugador.")
            break

        if respuesta_usuario == p["respuesta"]:
            print("✅ ¡Correcto!\n")
            acertadas += 1
        else:
            print(f"❌ Incorrecto. La respuesta correcta era {p['respuesta']}.\n")
            fallos += 1

        print(f"Aciertos: {acertadas} | Fallos: {fallos}")
        print("-" * 30)

        i += 1  # pasar a la siguiente pregunta

    # Mensajes finales según resultado
    if acertadas == 5:
        print("🏆 ¡Has ganado! Respondiste 5 preguntas correctamente.")
    elif fallos == 3:
        print("💀 Has perdido. Fallaste 3 veces.")
    else:
        print("🎉 Juego terminado. Se acabaron las preguntas.")

    print(f"Puntuación final: {acertadas} / {contestadas}")
```

```python
jugar_trivial()
```
